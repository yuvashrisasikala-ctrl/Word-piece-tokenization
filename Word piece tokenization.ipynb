```python
from collections import Counter

print("WordPiece Tokenizer Started")


```

```python
words = {
    "playing": 3,
    "played": 2,
    "player": 2
}

splits = {}

for word in words:
    tokens = [word[0]] + ["##" + c for c in word[1:]]
    splits[word] = tokens

print("Initial Splits:")
for word, tokens in splits.items():
    print(word, "->", tokens)
```

```python
token_freq = Counter()
pair_freq = Counter()

for word, freq in words.items():
    tokens = splits[word]

    for token in tokens:
        token_freq[token] += freq

    for i in range(len(tokens) - 1):
        pair = (tokens[i], tokens[i + 1])
        pair_freq[pair] += freq

print("Token Frequency:")
print(token_freq)

print("\nPair Frequency:")
print(pair_freq)
```

```python
scores = {}

for pair, freq in pair_freq.items():
    first, second = pair

    score = freq / (
        token_freq[first] * token_freq[second]
    )

    scores[pair] = score

print("WordPiece Scores:\n")

for pair, score in scores.items():
    print(pair, "=", round(score, 4))
    
    
```

```python
best_pair = max(scores, key=scores.get)

print("Best Pair:", best_pair)
print("Best Score:", round(scores[best_pair], 4))






```

```python
vocab = {
    "p",
    "##l",
    "##a",
    "##y",
    "##i",
    "##n",
    "##g",
    "##e",
    "##d",
    "##r",
    "play",
    "played",
    "player"
}

def tokenize(word):
    tokens = []

    while word:
        found = False

        for i in range(len(word), 0, -1):
            part = word[:i]

            if len(tokens) > 0:
                part = "##" + part

            if part in vocab:
                tokens.append(part)
                word = word[i:]
                found = True
                break

        if not found:
            return ["[UNK]"]

    return tokens


print("playing ->", tokenize("playing"))
print("played  ->", tokenize("played"))
print("player  ->", tokenize("player"))










```
